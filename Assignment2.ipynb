```python

class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species
        print(f"From inside class: {self}")

    def describe(self):
        print(f"I'm {self.name}, a {self.species}")

a1 = Animal("Bruno", "Dog")
print(f"From outside of class: {a1}")

a2 = Animal("Kitty", "Cat")
print(f"From outside of class: {a2}")


```

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
        print(f"From inside class: {self}")

    def (self):
        print(f"This is a {self.brand} {self.model}")

c1 = Car("Toyota", "Innova")
print(f"From outside of class: {c1}")
c2 = Car("Hyundai", "Creta")
print(f"From outside of class: {c2}")

    
```
